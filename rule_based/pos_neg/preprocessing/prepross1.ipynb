```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords as stpwrds
import pandas as pd
import string
import re
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

stopwords = nltk.corpus.stopwords.words("english")
```

```python
# Loading the data.

with open(r'C:\Users\timur\Documents\GitHub\EmailSentin\rule_based\pos_neg\preprocessing\example.txt', 'r') as file:
    # Read the entire content of the file into a string
    ex_text = file.read()
    
    

# replace multiple consecutive white space characters with a single space
def is_number(token):
        # Check if the token is a number (integer or decimal)
        return re.match(r'^\d+(\.\d+)?$', token) is not None

```

```python
ex_text
```

```python
# Remove the whitespaces.
ex_text = ex_text.strip()
ex_text = " ".join(ex_text.split())


# Tokenizing the text.
tokens = nltk.word_tokenize(ex_text)
# Lowercasing the tokes.
lowercased_tokens = [token.lower() for token in tokens]
# Filtering the punctuation and stop-words.
filtered_tokens = [token for token in lowercased_tokens if (token not in string.punctuation) and (token not in stopwords)]
# Removing numbers
filtered_tokens = [token for token in filtered_tokens if not is_number(token)]
```

```python
# Loading the positive-negative valence lexicons. 
file_path = r'C:\Users\timur\Documents\GitHub\EmailSentin\rule_based\pos_neg\lexicons\AFINN-111.txt'
with open(file_path, 'r', encoding='utf-8') as file:
    # Use a list comprehension to split each line and create tuples
    afinn111 = dict(map(lambda kv: (kv[0], int(kv[1])), 
                     [line.strip().split('\t') for line in file]))

print("AFINN-111 Lexicon loaded successfully with {} entries.".format(len(afinn)))

# Loading the positive-negative valence lexicons. 
file_path = r'C:\Users\timur\Documents\GitHub\EmailSentin\rule_based\pos_neg\lexicons\AFINN-96.txt'
with open(file_path, 'r', encoding='utf-8') as file:
    # Use a list comprehension to split each line and create tuples
    afinn96 = dict(map(lambda kv: (kv[0], int(kv[1])), 
                     [line.strip().split('\t') for line in file]))

print("AFINN-96 Lexicon loaded successfully with {} entries.".format(len(afinn)))

```

```python
afinn_96_score = sum(map(lambda word: afinn96.get(word, 0), filtered_tokens))
afinn_111_score = sum(map(lambda word: afinn111.get(word, 0), filtered_tokens))
```

```python
print(f'The Afinn-96 score of the text is: {afinn_96_score} \nThe Afinn-111 score of the text is: {afinn_111_score}')
```
