```python
import pandas as pd

def search_word_in_dataframe(df, word):
    # Function to check if the word is in the cell
    def word_in_cell(cell):
        if pd.isna(cell):  # Skip NaN values
            return False
        words = cell.split(',')
        return word in [w.strip() for w in words]

    # Apply the function to the entire DataFrame and get the rows where the word is found
    mask = df.applymap(word_in_cell).any(axis=1)
    return df.index[mask]

# Example usage
data = {
    'Column1': ['apple, banana, cherry', 'dog, cat', 'apple, dog'],
    'Column2': ['elephant, apple', 'banana, cherry', 'cat, dog']
}

df = pd.DataFrame(data)

word_to_search = 'apple'
result_indices = search_word_in_dataframe(df, word_to_search)
print(result_indices)
```
