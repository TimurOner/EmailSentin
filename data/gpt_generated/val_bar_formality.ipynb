```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Define values ranging from 0 to 0.3
values = np.linspace(0, 0.3, 256)

# Create a palette of 5 distinct colors transitioning from blue to white
discrete_colors = sns.light_palette("blue", n_colors=5, as_cmap=False)[::-1]  # Reverse the colors

# Create a colormap from the reversed discrete colors
cmap = sns.color_palette(discrete_colors)

# Plot the color gradient as a thinner and shorter vertical bar
fig, ax = plt.subplots(figsize=(0.4, 3))  # Thinner and shorter figure size
sns.heatmap(np.array([values]).T, cmap=sns.color_palette(discrete_colors), cbar=False, ax=ax)

# Set ticks with "Low" and "High" labels at the upper and lower edge
ax.set_yticks([0, 255])  # Set ticks only at the edges
ax.set_yticklabels(['High', 'Low'])  # Labels for ticks (reversed)
ax.set_xticks([])  # Hide x ticks

# Move y-ticks to the right
ax.yaxis.tick_right()
ax.yaxis.set_label_position("right")  # Position label on the right if you have one

# Rotate tick labels to be vertical
ax.tick_params(axis='y', rotation=0)  # Set the y-tick labels to be vertical

plt.show()
```
